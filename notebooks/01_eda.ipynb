# 01. Exploratory Data Analysis (EDA) — Air Quality Data in India
### AI for Sustainability Portfolio Project

This notebook explores the Kaggle *Air Quality Data in India (2015-2020)* dataset, focusing on Delhi, Mumbai, and Bengaluru to analyze pollutant dynamics, seasonal patterns, and autocorrelation.

```python
import os
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

# Set visual style
sns.set_theme(style="whitegrid")
plt.rcParams["figure.figsize"] = (12, 6)
```

## 1. Load Cleaned Dataset

```python
data_path = os.path.join("..", "data", "processed", "clean_data.csv")
df = pd.read_csv(data_path)
df["Date"] = pd.to_datetime(df["Date"])
print(f"Loaded {len(df)} records across cities: {df['City'].unique().tolist()}")
df.head()
```

## 2. City-wise Distribution of AQI

```python
plt.figure(figsize=(10, 5))
sns.boxplot(data=df, x="City", y="AQI", palette="Set2")
plt.title("AQI Distribution Across Key Indian Metros (2015-2020)", fontsize=14, fontweight="bold")
plt.ylabel("Air Quality Index (AQI)")
plt.show()
```

## 3. Seasonal Variation in Air Quality
Winter months experience temperature inversions in Northern India, trapping particulate matter.

```python
seasonal_aqi = df.groupby(["season", "City"])["AQI"].mean().unstack()
seasonal_aqi.plot(kind="bar", figsize=(11, 5), colormap="viridis")
plt.title("Mean AQI by Climatological Season and City", fontsize=13, fontweight="bold")
plt.ylabel("Average AQI")
plt.xlabel("Season")
plt.xticks(rotation=0)
plt.legend(title="City")
plt.show()
```

## 4. Pollutant Correlation Matrix

```python
pollutants = ["PM2.5", "PM10", "NO2", "SO2", "CO", "O3", "AQI"]
corr = df[pollutants].corr()

plt.figure(figsize=(8, 6))
sns.heatmap(corr, annot=True, fmt=".2f", cmap="coolwarm", vmin=-1, vmax=1, cbar=True)
plt.title("Correlation Matrix: Criteria Pollutants vs AQI", fontsize=13, fontweight="bold")
plt.show()
```
